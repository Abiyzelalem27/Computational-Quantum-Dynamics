# Split-step Fourier method

## Split-step Fourier method: Position and momentum space

We use units in which $\hbar = 1$ and $m = 1$.

### 1. Splitting the Hamiltonian

For a particle in one dimension,

$$
H = -\frac{1}{2}\frac{\partial^2}{\partial x^2}
    + V(x,t)
  = H_1 + H_2,
$$

where

$$
H_1 = -\frac{1}{2}\frac{\partial^2}{\partial x^2},
\qquad
H_2 = V(x,t).
$$

- The kinetic operator $H_1$ is diagonal in **momentum space**.
- The potential operator $H_2$ is diagonal in **position space**.

“Diagonal” means that the operator acts by multiplying each component
by its corresponding eigenvalue.

### 2. Plane waves are kinetic-energy eigenfunctions

Consider the plane wave

$$
\psi_p(x) = e^{ipx}.
$$

Its first and second derivatives are

$$
\frac{\partial}{\partial x}e^{ipx}
= ip\,e^{ipx},
$$

$$
\frac{\partial^2}{\partial x^2}e^{ipx}
= -p^2 e^{ipx}.
$$

Therefore,

$$
H_1 e^{ipx}
= -\frac{1}{2}\frac{\partial^2}{\partial x^2}e^{ipx}
= \frac{p^2}{2}e^{ipx}.
$$

Thus, a plane wave is an eigenfunction of $H_1$, with eigenvalue

$$
E_p = \frac{p^2}{2}.
$$

### 3. Fourier transform: Position to momentum

The Fourier transform expresses the same quantum state
in the momentum basis:

$$
\phi(p,t)
= \frac{1}{\sqrt{2\pi}}
  \int_{-\infty}^{\infty}
  e^{-ipx}\psi(x,t)\,dx.
$$

Here:

- $\psi(x,t)$ is the position-space wavefunction.
- $\phi(p,t)$ is the momentum-space wavefunction.
- $|\phi(p,t)|^2$ is the momentum probability density
  for a normalized state.

The inverse Fourier transform returns to position space:

$$
\psi(x,t)
= \frac{1}{\sqrt{2\pi}}
  \int_{-\infty}^{\infty}
  e^{ipx}\phi(p,t)\,dp.
$$

The Fourier transform changes the representation;
it does not itself advance the state in time.

### 4. Discrete position grid

For numerical calculations, take $N$ equally spaced points
in a periodic interval of length $L$.

The spacing is

$$
\Delta x = \frac{L}{N}.
$$

The positions are

$$
x_j = -\frac{L}{2} + j\Delta x,
\qquad j=0,1,\ldots,N-1.
$$

Thus, the grid is

$$
-\frac{L}{2},
\quad -\frac{L}{2}+\Delta x,
\quad \ldots,
\quad \frac{L}{2}-\Delta x.
$$

The endpoint $L/2$ is not repeated because periodic boundaries
identify it with $-L/2$.

A plane wave sampled on this grid has values

$$
\psi_j = e^{ipx_j}.
$$

### 5. Momentum spacing and smallest nonzero momentum

Periodic boundary conditions require

$$
e^{ip(x+L)} = e^{ipx}.
$$

Consequently,

$$
e^{ipL}=1
\quad\Longrightarrow\quad
pL=2\pi n,
\qquad n\in\mathbb{Z}.
$$

The allowed momenta are therefore

$$
p_n = n\frac{2\pi}{L}.
$$

The momentum spacing, which is also the smallest nonzero
momentum magnitude, is

$$
\boxed{\Delta p = \frac{2\pi}{L}}.
$$

Zero momentum is also allowed.
