## Lanczos Algorithm

The **Lanczos algorithm** approximates a few eigenpairs of a large Hermitian Hamiltonian $H$ without diagonalizing the full matrix.

An **eigenpair**(eigenenergy and eigenstate)  consists of an eigenenergy $E$ and its corresponding eigenstate $|\psi\rangle$:

$$
H|\psi\rangle=E|\psi\rangle.
$$

Lanczos is especially useful when the Hamiltonian is large and sparse.

## Krylov Subspace Methods


$$
|\Psi_0\rangle=\sum_{k=1}^{N}c_k|k\rangle,
\qquad
c_k=\langle k|\Psi_0\rangle.
$$

Imaginary-time evolution gives

$$
e^{-H\tau}|\Psi_0\rangle
=
\sum_{k=1}^{N}e^{-E_k\tau}c_k|k\rangle
=
e^{-E_1\tau}
\left(
c_1|1\rangle+
\sum_{k=2}^{N}e^{-(E_k-E_1)\tau}c_k|k\rangle
\right).
$$

Since $E_k-E_1>0$ for $k>1$, the excited-state terms become small **relative to the ground-state term** as $\tau\to\infty$. 

If $c_1\neq0$, the state approaches the ground state $|1\rangle$ **after normalization** (up to an overall phase).

In principle, we could find the ground state by repeatedly applying $e^{-H\Delta\tau}$ and normalizing after each step. However, calculating its action can be difficult, and convergence may be slow.

### Matrix Power Method

Repeated application of $H$ gives

$$
H^n|\Psi_0\rangle
=
\sum_k c_k\lambda_k^n|k\rangle
=
\lambda_{\max}^{\,n}
\left(
c_{\max}|k_{\max}\rangle+
\sum_{k\ne k_{\max}}
c_k\left(\frac{\lambda_k}{\lambda_{\max}}\right)^n|k\rangle
\right),
$$

where $|\lambda_{\max}|$ is the **largest eigenvalue magnitude**. If this magnitude is uniquely largest and $c_{\max}\neq0$, the other components decrease relative to $|k_{\max}\rangle$. After normalization, the vector approaches that eigenvector. It need not be the ground state.

### Lanczos and the Krylov Subspace

Lanczos uses the whole space generated by repeated applications of $H$:

$$
\mathcal K_n(H,|\Psi_0\rangle)
=
\operatorname{span}
\left\{
|\Psi_0\rangle,\,
H|\Psi_0\rangle,\,
H^2|\Psi_0\rangle,\,
\ldots,\,
H^{n-1}|\Psi_0\rangle
\right\}.
$$

It builds an orthonormal basis for this Krylov subspace, represents the Hermitian Hamiltonian $H$ as a smaller tridiagonal matrix, and diagonalizes that matrix to approximate eigenpairs.

Imaginary time filters toward the ground state. The power method illustrates repeated application of a matrix. Lanczos uses the resulting Krylov subspace to solve a smaller eigenvalue problem.
