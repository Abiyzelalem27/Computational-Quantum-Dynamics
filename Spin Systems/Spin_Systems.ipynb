# Spin Systems 

 
## $$\theta \in [0,\pi], \qquad \phi \in [0,2\pi)$$

```python
theta = np.pi / 2
phi = np.pi / 4

ket_0 = np.array([1, 0], dtype=complex)
ket_1 = np.array([0, 1], dtype=complex)

psi = (np.cos(theta / 2) * ket_0 + np.exp(1j * phi) * np.sin(theta / 2) * ket_1)

print("Qubit state=", psi)
print("Normalization=", np.vdot(psi, psi).real)
```

* **Operators:**

  ### Hermitian operators (Hamiltonians / observables)

  $$H = \vec{\alpha} \cdot \vec{\sigma} = \sum_{i=0}^{3} \alpha_i \sigma_i \quad (\sigma_0 = \mathbb{1}, \sigma_1 = \sigma_x, \sigma_2 = \sigma_y, \sigma_3 = \sigma_z)$$

Therefore, a general Hermitian operator can also be written as

$$
H=
\alpha_0 I
+\alpha_x\sigma_x
+\alpha_y\sigma_y
+\alpha_z\sigma_z.
$$

  * Pauli-operators

  * "spin operators" $S_i = \frac{1}{2} \sigma_i \quad (i=1,2,3)$

    $$\left(\text{or } \frac{\hbar}{2} \sigma_i\right)$$


  ---

  ### Unitary operators (evolution operators / rotations)

  $$U = e^{-i H t}$$

  > **Note:** $e^{-i \mathbb{1} t} |\psi\rangle = e^{-i t} |\psi\rangle$ is an irrelevant global phase $\rightarrow$ global energy shifts in $H$ are irrelevant.

  $$\rightarrow U = e^{-i t \sum_{i=1}^{3} n_i \sigma_i} \quad \text{with } |\vec{n}| = 1 \text{ parameterizes all relevant unitaries (3 parameters)}$$

  *(could also use $e^{i \vec{v} \cdot \vec{\sigma}}$ with general $\vec{v} \in \mathbb{R}^3$ (unnormalized), but we use $t, \vec{n}$ as it can be interpreted as an angle and rotation axis on the Bloch sphere as we will see)*

  $$\rightarrow \text{group of special } (\det U = 1) \text{ unitary group } SU(2)$$

  * Lie group with generators $\sigma_i \quad (i=1,2,3)$

  $$[\sigma_i, \sigma_j] = 2i \varepsilon_{ijk} \sigma_k \quad \text{Lie algebra}$$

### Solution:

$$U = e^{-i \Omega t S_x} = e^{-i \frac{\alpha}{2} \sigma_x} \quad (\alpha = \Omega t)$$

$$= \sum_{n} \frac{\left(-i \frac{\alpha}{2}\right)^n}{n!} \sigma_x^n \quad (\text{exploiting } \sigma_x^2 = \mathbb{1})$$

$$= \sum_{n \text{ even}} \frac{\left(-i \frac{\alpha}{2}\right)^n}{n!} \mathbb{1} + \sum_{n \text{ odd}} \frac{\left(-i \frac{\alpha}{2}\right)^n}{n!} \sigma_x$$

$$= \cos\left(\frac{\alpha}{2}\right) \mathbb{1} - i \sin\left(\frac{\alpha}{2}\right) \sigma_x$$

$$U |0\rangle = \cos\left(\frac{\Omega t}{2}\right) |0\rangle + e^{-i \frac{\pi}{2}} \sin\left(\frac{\Omega t}{2}\right) |1\rangle$$

$$\text{rotation around the } x\text{-axis in math. pos. sense.}$$

**Generally:** $e^{-i \alpha \vec{n} \cdot \vec{S}}$ is a rotation around axis $\vec{n}$ through an angle $\alpha$ on the Bloch sphere.
